```python
from tkinter import *
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
import google.generativeai as genai
```

```python
df = pd.read_csv('RTA.csv')
```

```python
window = Tk()
headingFormat = ('Kubo', 60, 'bold')
subheadingFormat = ('Cormorant Garamond', 16, 'normal')
buttonFormat = ('Cormorant Garamond', 20, 'bold')
buttonFormat2 = ('Cormorant Garamond', 16, 'bold')
bodyFormat = ('Cormorant Garamond', 12, 'bold')
```

```python
def destroy():
    for widgets in window.winfo_children():
        widgets.destroy()
```

```python
def endProgram():
    window.destroy()
```

```python
def getInsights(input):
    API_KEY = 0  # [API KEY HERE]

    genai.configure(api_key=API_KEY)

    model = genai.GenerativeModel('gemini-2.5-flash')
    chat = model.start_chat()

    defaultPrompt = 'Using the dataset provided, '
    finalPrompt = defaultPrompt + input

    data_sample = df.head(50).to_string(index=False)

    defaultPrompt = 'Using the dataset provided, '
    finalPrompt = defaultPrompt + input

    response = chat.send_message([finalPrompt, 'Here is the dataset sample:', data_sample])
    print(response.text)
```

```python
def showDataSet():
    destroy()
    dataset = Label(window, text=df, font=bodyFormat,
                    fg='#F4C430', bg='#2C2C34')
    dataset.place(x=0, y=0)
```

```python
def visualization2(graphStr):
    destroy()

    def generate_graph_scatter():
        ax.clear()
        selected_x = listbox_x.get(listbox_x.curselection())
        selected_y = listbox_y.get(listbox_y.curselection())

        ax.scatter(df[selected_x], df[selected_y], color='#F4C430')
        fig.patch.set_facecolor('#2F2C34')
        ax.set_facecolor('#2F2C34')
        ax.tick_params(axis='x', colors='#F4C430', rotation=45)
        ax.tick_params(axis='y', colors='#F4C430')
        ax.set_xlabel(selected_x, color='#F4C430')
        ax.set_ylabel(selected_y, color='#F4C430')
        fig.tight_layout()
        canvas.draw()
        
    def generate_graph_histo():
        ax.clear()
        selected_x = listbox_x.get(listbox_x.curselection())

        ax.hist(df[selected_x], color='#F4C430')
        fig.patch.set_facecolor('#2F2C34')
        ax.set_facecolor('#2F2C34')
        ax.tick_params(axis='x', colors='#F4C430', rotation=45)
        ax.tick_params(axis='y', colors='#F4C430')
        ax.set_xlabel(selected_x, color='#F4C430')
        fig.tight_layout()
        canvas.draw()

    def generate_graph_line():
        ax.clear()
        selected_x = listbox_x.get(listbox_x.curselection())
        selected_y = listbox_y.get(listbox_y.curselection())

        if (selected_x == 'Time' or selected_x == 'Age Range'):
            ax.plot(df[selected_x].mean(), df[selected_y], color='#F4C430')
            fig.patch.set_facecolor('#2F2C34')
            ax.set_facecolor('#2F2C34')
            ax.tick_params(axis='x', colors='#F4C430', rotation=45)
            ax.tick_params(axis='y', colors='#F4C430')
            ax.set_xlabel(selected_x, color='#F4C430')
            ax.set_ylabel(selected_y, color='#F4C430')
            fig.tight_layout()
            canvas.draw()
        else:
            ax.plot(df[selected_x], df[selected_y], color='#F4C430')
            fig.patch.set_facecolor('#2F2C34')
            ax.set_facecolor('#2F2C34')
            ax.tick_params(axis='x', colors='#F4C430', rotation=45)
            ax.tick_params(axis='y', colors='#F4C430')
            ax.set_xlabel(selected_x, color='#F4C430')
            ax.set_ylabel(selected_y, color='#F4C430')
            fig.tight_layout()
            canvas.draw()

    def generate_graph_pie():
        ax.clear()
        selected_x = listbox_x.get(listbox_x.curselection())

        values = df[selected_x].value_counts()
    
        colors = ['#624E13', '#AB8922', '#F4C430', '#F7D66E', '#FBE7AC']
    
        wedges, texts = ax.pie(values, labels=None, colors=colors, rotatelabels=True)

        leg = ax.legend(wedges, values.index, title=selected_x,
                            loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))
            
        plt.setp(leg.get_texts(), color='#F4C430')
        plt.setp(leg.get_title(), color='#F4C430')
        frame = leg.get_frame()
        frame.set_facecolor('#2F2C34') 
        frame.set_edgecolor('#2F2C34') 

        fig.patch.set_facecolor('#2F2C34') 
        ax.set_facecolor('#2F2C34')        
        canvas.draw() 

    backBtn = Button(window, text='Back', command=visualization, font=buttonFormat2,
                     fg='#F4C430', bg='#2F2C34',
                     activebackground='#2C2C34', activeforeground='#D7263D')
    backBtn.config(width=7)
    backBtn.place(x=700, y=530)

    if (graphStr == 'Scatter Graph'):

        dictX = {1: 'Time', 2: 'Age Range', 3: 'Month', 4: 'Year'}
        dictY = {1: 'responsetime', 2: 'ACCLASS Severity', 3: 'Vehicle Count'}

        heading = Label(window, text=graphStr, font=headingFormat,
                        fg='#F4C430', bg='#2F2C34') 
        heading.pack(pady=10)

        listbox_x = Listbox(window, font=buttonFormat, selectmode='single', exportselection=0,
                            fg='#F4C430', bg='#2F2C34')
        listbox_x.config(height=listbox_x.size())
        listbox_x.place(x=550, y=100)
        for key, value in dictX.items():
            listbox_x.insert(key, value)

        listbox_y = Listbox(window, font=buttonFormat, selectmode='single', exportselection=0,
                            fg='#F4C430', bg='#2F2C34')
        listbox_y.config(height=listbox_y.size())
        listbox_y.place(x=550, y=335)
        for key, value in dictY.items():
            listbox_y.insert(key, value)

        submitBtn = Button(window, text='Generate Graph', command=generate_graph_scatter, font=buttonFormat2,
                           fg='#F4C430', bg='#2F2C34', 
                           activebackground='#2C2C34', activeforeground='#D7263D')
        submitBtn.config(width=7)
        submitBtn.place(x=600, y=530)

        fig, ax = plt.subplots(figsize=(5, 4))
        canvas = FigureCanvasTkAgg(fig, master=window)
        canvas.get_tk_widget().place(x=20, y=100)

        '''*********************************************************************'''
    elif (graphStr == 'Histogram'):
        dictX = {1: 'Year', 2: 'Month', 3: 'Date', 4: 'Time', 5: 'Age Range',
             6: 'responsetime', 7: 'ACCLASS Severity'}

        heading = Label(window, text=graphStr, font=headingFormat,
                        fg='#F4C430', bg='#2F2C34') 
        heading.pack(pady=10)

        listbox_x = Listbox(window, font=buttonFormat,
                            fg='#F4C430', bg='#2F2C34')
        listbox_x.config(height=listbox_x.size())
        listbox_x.place(x=550, y=100)
        for key, value in dictX.items():
            listbox_x.insert(key, value)

        submitBtn = Button(window, text='Generate Graph', command=generate_graph_histo, font=buttonFormat2,
                           fg='#F4C430', bg='#2F2C34', 
                           activebackground='#2C2C34', activeforeground='#D7263D')
        submitBtn.config(width=7)
        submitBtn.place(x=600, y=530)

        fig, ax = plt.subplots(figsize=(5, 4))
        canvas = FigureCanvasTkAgg(fig, master=window)
        canvas.get_tk_widget().place(x=20, y=100)

        '''*********************************************************************'''
    elif (graphStr == 'Line Graph'):
        dictX = {1: 'Year', 2: 'Month', 3: 'Date', 4: 'Time', 5: 'InjuryType', 6: 'Age Range'}
        dictY = {1: 'ACCLASS Severity', 2: 'responsetime', 3: 'PatientStatus'}

        heading = Label(window, text=graphStr, font=headingFormat,
                        fg='#F4C430', bg='#2F2C34') 
        heading.pack(pady=10)

        listbox_x = Listbox(window, font=buttonFormat, selectmode='single', exportselection=0,
                            fg='#F4C430', bg='#2F2C34')
        listbox_x.config(height=listbox_x.size())
        listbox_x.place(x=550, y=100)
        for key, value in dictX.items():
            listbox_x.insert(key, value)

        listbox_y = Listbox(window, font=buttonFormat, selectmode='single', exportselection=0,
                            fg='#F4C430', bg='#2F2C34')
        listbox_y.config(height=listbox_y.size())
        listbox_y.place(x=550, y=335)
        for key, value in dictY.items():
            listbox_y.insert(key, value)
    
        submitBtn = Button(window, text='Generate Graph', command=generate_graph_line, font=buttonFormat2,
                           fg='#F4C430', bg='#2F2C34', 
                           activebackground='#2C2C34', activeforeground='#D7263D')
        submitBtn.config(width=7)
        submitBtn.place(x=600, y=530)

        fig, ax = plt.subplots(figsize=(5, 4))
        canvas = FigureCanvasTkAgg(fig, master=window)
        canvas.get_tk_widget().place(x=10, y=100)

        '''*********************************************************************'''
    elif (graphStr == 'Pie Chart'):
        dictX = {1: 'Season', 2: 'Light', 3: 'Weather', 4: 'Gender', 5: 'Age Range', 6: 'EducationTitle', 7: 'ACCLASS Severity', 8: 'InjuryType',
        9: 'Cause', 10: 'PatientStatus', 11: 'Vehicles Involved'}

        heading = Label(window, text=graphStr, font=headingFormat,
                        fg='#F4C430', bg='#2F2C34') 
        heading.pack(pady=10)

        listbox_x = Listbox(window, font=buttonFormat,
                            fg='#F4C430', bg='#2F2C34')
        listbox_x.config(height=listbox_x.size())
        listbox_x.place(x=550, y=100)
        for key, value in dictX.items():
            listbox_x.insert(key, value)
    
        submitBtn = Button(window, text='Generate Graph', command=generate_graph_pie, font=buttonFormat2,
                           fg='#F4C430', bg='#2F2C34', 
                           activebackground='#2C2C34', activeforeground='#D7263D')
        submitBtn.config(width=7)
        submitBtn.place(x=600, y=530)

    if graphStr != 'Pie Chart':
        fig, ax = plt.subplots(figsize=(5, 4))
        canvas = FigureCanvasTkAgg(fig, master=window)
        canvas.get_tk_widget().place(x=20, y=100)
    else:
        fig, ax = plt.subplots(figsize=(6, 4))
        canvas = FigureCanvasTkAgg(fig, master=window)
        canvas.get_tk_widget().place(x=-80, y=100)


'''*******************************************************************'''

def visualization():

    def graphType():
        graphStr = graphTypeList.get(graphTypeList.curselection())
        visualization2(graphStr)

    destroy()

    exitBtn = Button(window, text='Exit', command=endProgram, font=buttonFormat2,
                     fg='#F4C430', bg='#2C2C34', cursor='Pirate',
                     activebackground='#2C2C34', activeforeground='#D7263D')
    exitBtn.config(width=7)
    exitBtn.place(x=700, y=530)

    graphList = {1: 'Scatter Graph', 2: 'Histogram', 3: 'Line Graph', 4: 'Pie Chart'}

    graphTypeList = Listbox(window, font=buttonFormat,
                            fg='#F4C430', bg='#2F2C34')
    graphTypeList.config(height=graphTypeList.size())
    graphTypeList.pack(pady=200)
    for key, value in graphList.items():
        graphTypeList.insert(key, value)

    submitBtn = Button(window, text='Generate Graph', command=graphType, font=buttonFormat,
                       fg='#F4C430', bg='#2F2C34')
    submitBtn.config(width=18)
    submitBtn.place(x=278, y=370)
```

```python
def geminaiAPI():
    destroy()

    def submit(event = None):
        input = userInput.get("1.0", "end-1c")
        print(input)
        getInsights(input)
        return 'break'

    userInput = Text(height=100, width=100)
    userInput.config(font=bodyFormat, fg='#F4F430', bg='#2C2C34', 
                     border=3)
    userInput.place(x=50, y=50, height=200)
    userInput.bind('<Return>', submit)

    submitBtn = Button(window, text='Submit', command=submit, font=buttonFormat,
                       fg='#F4C430', bg='#2C2C34')
    submitBtn.place(x=50, y=250)
```

```python
df.drop(['EcNumber', 'Nature of Weekday', 'CallTime', 'TotalPatientsInEmergency'], axis=1, inplace=True)

vehicle_cols = ['BicycleInvovled', 'BikesInvolved', 'BusesInvolved', 'CarsInvolved', 'CartInvovled', 'RickshawsInvolved', 'TractorInvovled',
            'TrainsInvovled', 'TrucksInvolved', 'VansInvolved', 'OthersInvolved'] 
clean_names = ['Bicycle', 'Bike', 'Buses', 'Cars', 'Carts', 'Rickshaws', 'Tractor', 'Trains', 'Trucks', 'Vans', 'Others'] # corresponding clean names

mask = df[vehicle_cols] > 0
mask.columns = clean_names

df['Vehicles Involved'] = mask.dot(mask.columns + ', ').str.rstrip(', ')
df['Vehicles Involved'] = df['Vehicles Involved'].replace('', 'None')

df.drop(vehicle_cols, axis=1, inplace=True)
df['responsetime'].fillna(df['responsetime'].mean(), inplace=True)


severity_map = {'Fatal': 5, 'Multiple Fracture': 4, 'Single Fracture': 3,
            'Minor': 2, 'Non-Fatal': 1}
df['ACCLASS Severity'] = df['ACCLASS'].map(severity_map)


df['Vehicles Count'] = df['Vehicles Involved'].apply(lambda x: len(x.split(",")))
```
